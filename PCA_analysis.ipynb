```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import pandas as pd

genomic_data = pd.read_csv(genomic_file)

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
genomic_data_normalized = scaler.fit_transform(genomic_data)

pca = PCA()
pca.fit(genomic_data_normalized)

cumulative_variance = pca.explained_variance_ratio_.cumsum()

plt.figure(figsize=(8, 6))
plt.plot(cumulative_variance, marker='o', linestyle='--')
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Explained Variance vs Number of Components")
plt.grid()
plt.show()

optimal_components_90 = (cumulative_variance >= 0.90).argmax() + 1
optimal_components_95 = (cumulative_variance >= 0.95).argmax() + 1

print(f"Number of components for 90% variance: {optimal_components_90}")
print(f"Number of components for 95% variance: {optimal_components_95}")
```
